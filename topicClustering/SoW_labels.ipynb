# About this notebook
This notebook takes the mission assignments file and merges it with the LLM generated SoW list.  This notebook must be run prior to using the SoW topics notebook as it generates the llm_ma_merge and filtered_topics csv files.

```python
import pandas as pd
```

# Merge the LLM output with the original mission assignments dataset

```python
df_MA = pd.read_csv('MissionAssignments.csv')
```

```python
len(df_MA)
```

```python
#import the file generated by LLM
df_llm=pd.read_csv('actions_topics_bedrock_full_sow.csv')
```

```python
#len(df_llm)
```

```python
combined_df=pd.merge(df_llm,df_MA,how='left',on='statementOfWork')
```

```python
#len(combined_df)
```

```python
combined_df.to_csv('llm_MA_merge.csv')
```

# Create csv file with the topics filtered

```python
reduced=combined_df[['maId','clean_statement_topic','declarationType','maAmendNumber','supportFunction']]
```

```python
filtered=reduced[(reduced['declarationType']!='SU')&(reduced['maAmendNumber']==0)&(reduced['supportFunction']<=15)]
```

```python
len(filtered)
```

```python
top_list=filtered['clean_statement_topic'].unique().tolist()
```

```python
#len(top_list)
```

```python
filtered.dropna(inplace=True)
```

```python
filtered.head()
```

```python
filtered.to_csv('filtered_topics.csv')
```

# Combined AR and filtered topic csv generation
This was not used in the final model, but the code is preserved as it may be useful for future analysis and updates.

```python
filtered['sow_top_num']=0
```

```python
filtered.head()
```

```python
filtered.reset_index(inplace=True)
```

```python
for i in range(len(filtered)):
    if filtered['clean_statement_topic'].iloc[i] in top_list:
        filtered['sow_top_num'].iloc[i]=top_list.index(filtered['clean_statement_topic'].iloc[i])
```

```python
filtered.head(20)
```

```python
export=filtered[['maId','sow_top_num']]
```

```python
export.to_csv('SoW_topic_numbers.csv')
```

```python
topics_and_nums=filtered[['clean_statement_topic','sow_top_num']]
topics_and_nums.drop_duplicates(inplace=True)
```

```python
len(export)
```

```python
topics_and_nums.to_csv('sow_topics_with_nums.csv')
```

```python
AR_df=pd.read_csv('AR_topics.csv')
```

```python
combined_topics=pd.merge(export,AR_df,how='left',on='maId')
```

```python
len(combined_topics)
```

```python
combined_topics.head()
```

```python
combined_export=combined_topics[['maId','sow_top_num','AR_topic']]
```

```python
combined_export.to_csv('sow_AR_top_nums.csv')
```
